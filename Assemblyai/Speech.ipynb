```python
import assemblyai as aai
from docx import Document
import json
import os
from pathlib import Path
import time
import sys


# Load API key from file
def load_api_key(path=r"C:\APIkey\assemblyai.key"):
    with open(path, "r") as f:
        return f.read().strip()

# Set the API key for AssemblyAI
aai.settings.api_key = load_api_key()
```

```python
# List your custom phrases here
custom_phrases = [
    "کاستاندا",    "سالک",    "شامن",    "سالک مبارز",    "فلوریندا",    "هوده",    "دون خوان",
    "هوده تری",    "دون خنارو",    "خودآگاه",    "تولتک",    "نا خودآگاه",    "تولنکهای کهن",    "تهیگی",
    "دقت اول",    "دقت دوم"
]

# custom_phrases = [{
#                     "value": "کاستاندا","boost": 10.0}, {"value": "سالک","boost": 10.0},
#                     {"value": "شامن","boost": 5.0}, {"value": "سالک مبارز","boost":15.0},
#                     {"value": "فلوریندا","boost": 15.0}, {"value": "هوده","boost": 15.0},
#                     {"value": "دون خوان","boost": 20.0}, {"value": "هوده تری","boost": 5.0},
#                     {"value": "دون خنارو","boost": 20.0}, {"value": "خودآگاه","boost": 10.0},
#                     {"value": "تولتک","boost": 15.0}, {"value": "نا خودآگاه","boost": 5.0},
#                     {"value": "تولنکهای کهن","boost": 12.0}, {"value": "تهیگی","boost": 10.0},
#                     {"value": "دقت اول","boost": 12.0},
#                     {"value": "دقت دوم","boost": 12.0}]

# Configure transcription
config = aai.TranscriptionConfig(
    speaker_labels=True,
    language_code="fa",
    word_boost=custom_phrases
)
```

```python
def process_file(file_path):
    try:
        transcriber = aai.Transcriber()
        transcript = transcriber.transcribe(str(file_path), config)   
           
        print(f"File {file_path.stem} - Transcript ID: {transcript.id}")
        
        if transcript.error:
            print(f"Error in {file_path.stem}: {transcript.error}")
            return False
            
        # Create document
        doc = Document()
        doc.add_heading(f'Transcript - {file_path.stem}', 0)
        
        # Add transcribed text
        for utterance in transcript.utterances:
            doc.add_paragraph(f"Speaker {utterance.speaker}: {utterance.text}")
        
        # Save docx with same name as audio file
        docx_path = file_path.with_suffix('.docx')
        doc.save(docx_path)
        
        # Save JSON backup
        json_path = file_path.with_suffix('.json')
        with open(json_path, 'w', encoding='utf8') as json_file:
            json_str = json.dumps(transcript.json_response, ensure_ascii=False, indent=2)
            json_file.write(json_str)
            
        print(f"Successfully processed {file_path.stem}")
        return True
        
    except Exception as e:
        print(f"Error processing {file_path.stem}: {str(e)}")
        return False
```

```python

# Process files until first missing file
for i in range(1, 101):
    file_name = f"S ({i})"
    file_found = False
    
    # Check for file with any of the supported extensions
    for ext in ['.mp3', '.wav', '.m4a', '.mp4']:
        file_path = Path(file_name + ext)
        if file_path.exists():
            file_found = True
            print(f"\nProcessing {file_path}...")
            success = process_file(file_path)
            if success:
                time.sleep(2)  # Small delay between files
            break
    
    # Exit if no file found with any extension
    if not file_found:
        print(f"\nFile {file_name} not found with any supported extension. Stopping process.")
        sys.exit(0)
```
